**Findings**: `df_benign` (from `behaviors_benign.csv`) and `df_harmful` (from `behaviors_harmful.csv`) share the same schema - 100 rows × 6 columns each, no missing values in either:
- `Index` - integer row index (0–99)
- `Goal` - the request text itself; this is the input/text column
- `Target` - an affirmative response starter (e.g. "Sure, here is...") matching the Goal; not the model's actual output
- `Behavior` - a short, unique label for the specific topic (100 unique values per file, one per row)
- `Category` - the broad harm category the topic belongs to (10 categories, 10 rows each, e.g. Harassment/Discrimination, Malware/Hacking)
- `Source` - provenance of the behavior

Each row in `df_benign` is paired one-to-one with a row in `df_harmful` sharing the same `Behavior`/`Category`, written to look similar on the surface (same topic) but asking for something legitimate instead of harmful.

Differences between the two:
- `Goal`/`Target` content: `df_benign` asks for legitimate content, `df_harmful` asks for harmful content on the same topics
- `Source`: `df_benign` is entirely `Original` (100/100); `df_harmful` is mixed — `Original` (55), `TDC/HarmBench` (27), `AdvBench` (18)

**No toxicity information**: neither file has a toxicity score, toxicity label, or any risk/safety annotation. `Category`/`Behavior` describe topic, not risk, and "harmful" here only means which file/split a row came from, not a measured toxicity value. Many `df_harmful` rows ask for something dangerous or illegal without using offensive language (e.g. malware, fraud instructions), so "harmful" and "toxic" are not the same thing for this dataset — treat any toxicity claim about either dataframe as an assumption, not something present in the data.

#### Judge Comparison Test

```python
print(f"Number of Rows: {df_judge_comparisons.shape[0]}")
print(f"Number of Columns: {df_judge_comparisons.shape[1]}")
df_judge_comparisons.head(10)
```

```python
print ('----------------- Judge Comparison Data -----------------')
print (f"Number of missing values in df_judge_comparisons: {df_judge_comparisons.isnull().sum().sum()}")
print (f'Rows where prompt equals goal (no attack wrapper) in df_judge_comparisons: {(df_judge_comparisons["prompt"].str.strip() == df_judge_comparisons["goal"].str.strip()).sum()}')
print (f'0 = not harmful; 1 - harmful ')
print (f'Human majority label counts in df_judge_comparisons: {df_judge_comparisons["human_majority"].value_counts().to_dict()}')
print (f'Automated judge classifier counts in df_judge_comparisons: {({col: df_judge_comparisons[col].value_counts().to_dict() for col in ["harmbench_cf", "gpt4_cf", "llamaguard2_cf", "llama3_cf"]})}')
print()
print(df_judge_comparisons.info())
```

**Findings**: `df_judge` (from `judge_comparison_test.csv`), 300 rows × 12 columns, no missing values. Different schema from `df_benign`/`df_harmful`:
- `Index` - integer row index (0–299), all unique
- `goal` - the underlying harmful behavior being targeted (same style of text as `Goal` in `df_harmful`)
- `prompt` - the text actually sent to the model; for 100/300 rows this is identical to `goal` (a plain request), for the other 200 it's `goal` wrapped in an attack template (role-play, fake pretext, etc.) — i.e. this file, unlike the behaviors files, contains real jailbreak-style attack prompts
- `target_response` - the model's actual reply to `prompt` (not an affirmative stub like `Target` in the behaviors files)
- `human1`, `human2`, `human3` - three human annotators' binary verdict (0/1) on whether `target_response` was harmful (0 = not harmful, 1 = harmful)
- `human_majority` - majority vote of the three (190 rows `0`, 110 rows `1`)
- `harmbench_cf`, `gpt4_cf`, `llamaguard2_cf`, `llama3_cf` - four automated judge classifiers' binary verdicts on the same question, for comparing automated judges against the human majority

This file exists to benchmark response-harmfulness judges against human labels, not to classify prompts.

**No toxicity information for the prompt**: every human/automated label here scores `target_response` (the model's output), not `prompt` or `goal`. A refusal response scores 0 regardless of how risky the prompt was, so none of these columns can be used as a toxicity or jailbreak label for the input prompt. The `prompt` column itself is useful as unlabeled jailbreak-style text, but assigning it a risk label from this file would be an assumption, not something the data provides.

```python
base = os.path.join('..', 'data')
path = os.path.join(base, 'lmsys_toxicchat.zip')
with zipfile.ZipFile(path) as z:
    print(f'lmsys_toxicchat.zip contains:')
    for name in z.namelist():
        print(f'  {name}')
    print()
```

**Findings**: Toxic Chat Dataset consist of 2 csv files: 
1) train.csv

```python
with zipfile.ZipFile(os.path.join(base, 'lmsys_toxicchat.zip')) as z:
    with z.open('lmsys_toxicchat/train.csv') as f:
        df_toxic_train = pd.read_csv(f)
    with z.open('lmsys_toxicchat/test.csv') as f:
        df_toxic_test = pd.read_csv(f)
```

```python
print(f"Number of Rows: {df_toxic_test.shape[0]}")
print(f"Number of Columns: {df_toxic_test.shape[1]}")
df_toxic_test.head(10)
```

```python
print ('----------------- Test Data -----------------')
print (f"Number of missing values in df_toxic_test: {df_toxic_test.isnull().sum().sum()}")
print (f'Toxicity label counts in df_toxic_test: {df_toxic_test["toxicity"].value_counts().to_dict()}')
print (f'Jailbreaking label counts in df_toxic_test: {df_toxic_test["jailbreaking"].value_counts().to_dict()}')
print (f'Human annotation counts in df_toxic_test: {df_toxic_test["human_annotation"].value_counts().to_dict()}')
print()
print(df_toxic_test.info())
```

**Findings**: `df_toxic_train` (from `train.csv`) and `df_toxic_test` (from `test.csv`) share the same schema - 7 columns each, no missing values in either:
- `conv_id` - unique conversation identifier
- `user_input` - the user's prompt text; this is the input/text column
- `model_output` - the model's actual reply to `user_input`
- `human_annotation` - whether a human reviewed this row (`True`/`False`); train: 2,801 `True` / 2,281 `False`, test: 2,853 `True` / 2,230 `False`
- `toxicity` - binary label (0/1): whether `user_input` was judged toxic
- `jailbreaking` - binary label (0/1): whether `user_input` was a jailbreak attempt. Every `jailbreaking=1` row also has `toxicity=1` in both splits (jailbreak is a subset of toxic), so a 3-class label (safe/toxic/jailbreak) needs to give `jailbreaking` precedence over `toxicity`
- `openai_moderation` - raw OpenAI moderation API scores per category, stored as a string

Row counts: `df_toxic_train` 5,082 rows (4,698 safe, 271 toxic-only, 113 jailbreak); `df_toxic_test` 5,083 rows (4,721 safe, 271 toxic-only, 91 jailbreak).

**Unlike the JBB files, this dataset *does* have toxicity information**: `toxicity` and `jailbreaking` are real risk labels for the prompt itself (not the model's response), making Toxic Chat the only source in this project with both a toxicity label and a jailbreak label. `human_annotation=False` does not mean the label is missing — the released `toxicity`/`jailbreaking` values are still present; it only marks that a human did not personally review that particular row.
